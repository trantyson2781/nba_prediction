### Kelly stake example
Feed a **calibrated** model probability and the book's decimal odds.

```python
from nba_pred.betting import kelly

model_prob = 0.68   # calibrated P(home win) from predict_season
odds = 1.90         # book decimal odds
print("edge vs market:", round(kelly.edge(model_prob, odds), 3))
print("suggested stake:", round(kelly.suggest_bet_amount(model_prob, odds), 2))
```
