```python
import pandas as pd
import numpy as np
from nba_api.stats.endpoints import leaguegamefinder,boxscoretraditionalv2,leaguedashplayerstats,leaguegamelog
print('Set up completed!')
```

```python
from nba_api.stats.endpoints import leaguegamefinder
import pandas as pd

seasons = ["2021-22", "2022-23", "2023-24"]
all_games = []

for season in seasons:
    finder = leaguegamefinder.LeagueGameFinder(
        season_nullable=season,
        league_id_nullable="00",       # NBA league ID
        season_type_nullable="Regular Season"
    )
    games = finder.get_data_frames()[0]
    games['SEASON'] = season   # optional, keep track of season
    all_games.append(games)

# Combine all three seasons into one DataFrame
games_allseasons = pd.concat(all_games, ignore_index=True)

# Inspect
print(games_allseasons.columns)
print(games_allseasons.head(10))
```

```python
all_stars_list = []
all_star_attendance = []
for season in seasons:
    # Get player stats
    players_stats = leaguedashplayerstats.LeagueDashPlayerStats(season=season).get_data_frames()[0]
    stars = players_stats.sort_values('PTS', ascending=False).drop_duplicates('TEAM_ID')
    stars_list = stars[['TEAM_ID', 'PLAYER_ID', 'PLAYER_NAME']]
    stars_list['SEASON'] = season
    all_stars_list.append(stars_list)

    # Get player logs
    player_logs = leaguegamelog.LeagueGameLog(season=season, player_or_team_abbreviation='P').get_data_frames()[0]
    star_attendance = player_logs[player_logs['PLAYER_ID'].isin(stars_list['PLAYER_ID'])][['GAME_ID', 'PLAYER_ID']]
    star_attendance['IS_ACTIVE'] = 1
    star_attendance['SEASON'] = season
    all_star_attendance.append(star_attendance)

# Combine across seasons
stars_list_3seasons = pd.concat(all_stars_list, ignore_index=True)
star_attendance_3seasons = pd.concat(all_star_attendance, ignore_index=True)

# Inspect
print(stars_list_3seasons.head())
print(star_attendance_3seasons.head())
```

```python
games_allseasons['POSS'] = 0.96*(games_allseasons['FGA']+games_allseasons['TOV']+0.4*games_allseasons['FTA']-games_allseasons['OREB'])
games_allseasons['eFG%'] = (games_allseasons['FGM']+.5*games_allseasons['FG3M'])/games_allseasons['FGA']
games_allseasons['TS%'] = games_allseasons['PTS']/(2*games_allseasons['FGA']+0.44*games_allseasons['FTA'])
games_allseasons['NET_RAT'] = 100*(games_allseasons['PTS']/(games_allseasons['POSS'])-(games_allseasons['PTS']-games_allseasons['PLUS_MINUS'])/(games_allseasons['POSS']))
games_allseasons.head()
```

```python
games_allseasons['GAME_DATE'] = pd.to_datetime(games_allseasons['GAME_DATE'])
games_allseasons['CUM_WINS'] = games_allseasons.groupby('TEAM_ID')['WL'].transform(lambda x: (x == 'W').astype(int).cumsum().shift(1).fillna(0))
games_allseasons['GAMES_PLAYED'] = games_allseasons.groupby('TEAM_ID').cumcount().astype(int)
games_allseasons['SEASON_WIN_PCT'] = games_allseasons['CUM_WINS'] / games_allseasons['GAMES_PLAYED']
games_allseasons['SEASON_WIN_PCT'] = games_allseasons['SEASON_WIN_PCT'].fillna(0)
games_allseasons['DAYS_REST'] = games_allseasons.groupby('TEAM_ID')['GAME_DATE'].diff().dt.days
games_allseasons['DAYS_REST'] = games_allseasons['DAYS_REST'].clip(upper=4).fillna(4)
games_allseasons['IS_B2B'] = (games_allseasons['DAYS_REST'] <= 1).astype(int)
games_allseasons['IS_B2B'] = games_allseasons['IS_B2B'].fillna(0)

games_allseasons = games_allseasons.dropna()
games_allseasons = games_allseasons.sort_values(by=['TEAM_ID', 'GAME_DATE'])
cols_to_roll = ['AST', 'REB', 'TOV','TS%','NET_RAT','SEASON_WIN_PCT']
for col in cols_to_roll:
    games_allseasons['ROLL_' + col] = games_allseasons.groupby('TEAM_ID')[col].transform(lambda x: x.ewm(span=10).mean().shift(1))

from nba_api.stats.static import teams
team_abbr_to_id = {t['abbreviation']: t['id'] for t in teams.get_teams()}

games_allseasons['OPP_ABBR'] = games_allseasons['MATCHUP'].apply(lambda x: x.split(' ')[-1])
games_allseasons['OPP_ID'] = games_allseasons['OPP_ABBR'].map(team_abbr_to_id)

win_pct_map = games_allseasons.set_index(['TEAM_ID', 'GAME_DATE'])['SEASON_WIN_PCT'].to_dict()

games_allseasons['OPP_WIN_PCT_AT_GAME'] = games_allseasons.apply(
    lambda x: win_pct_map.get((x['OPP_ID'], x['GAME_DATE']), 0.5), axis=1
)

games_allseasons['ROLL_SOS'] = games_allseasons.groupby('TEAM_ID')['OPP_WIN_PCT_AT_GAME'].transform(
    lambda x: x.rolling(10).mean().shift(1).fillna(0.5)
)

games_allseasons['ADJ_NET_RAT'] = games_allseasons['ROLL_NET_RAT'] * games_allseasons['ROLL_SOS']

print(games_allseasons[:11])
games_allseasons.columns
```

```python
# 1. Clear the duplicate index issues by resetting
games_allseasons = games_allseasons.sort_values(['TEAM_ID', 'GAME_DATE']).reset_index(drop=True)

# 2. Run the rolling logic using the unique index
# include_groups=False ensures future-proofing
games_allseasons['GAMES_IN_LAST_4_DAYS'] = games_allseasons.groupby('TEAM_ID').apply(
    lambda x: x.set_index('GAME_DATE')['WL'].rolling('4D').count().shift(1).fillna(0),
    include_groups=False
).values # Use .values to ignore index alignment entirely

games_allseasons['GAMES_IN_LAST_5_DAYS'] = games_allseasons.groupby('TEAM_ID').apply(
    lambda x: x.set_index('GAME_DATE')['WL'].rolling('5D').count().shift(1).fillna(0),
    include_groups=False
).values

# 3. Create Flags
games_allseasons['FATIGUE_3_IN_4'] = (games_allseasons['GAMES_IN_LAST_4_DAYS'] >= 2).astype(int)
games_allseasons['FATIGUE_4_IN_5'] = (games_allseasons['GAMES_IN_LAST_5_DAYS'] >= 3).astype(int)
```

```python
stats_cols = ['GAME_ID','TEAM_NAME','TEAM_ID','WL','SEASON_WIN_PCT','DAYS_REST','GAMES_PLAYED','GAME_DATE','TS%','NET_RAT','IS_B2B','FATIGUE_3_IN_4','FATIGUE_4_IN_5','GAMES_IN_LAST_4_DAYS','GAMES_IN_LAST_5_DAYS','ADJ_NET_RAT'] + [col for col in games_allseasons.columns if 'ROLL_' in col]


home_games = games_allseasons[games_allseasons['MATCHUP'].str.contains('vs')][stats_cols].copy()
away_games = games_allseasons[games_allseasons['MATCHUP'].str.contains(' @ ')][stats_cols].copy()

home_games = home_games.add_prefix('HOME_').rename(columns={'HOME_GAME_ID': 'GAME_ID','HOME_GAME_DATE': 'GAME_DATE'})
away_games = away_games.add_prefix('AWAY_').rename(columns={'AWAY_GAME_ID': 'GAME_ID'})

model_df = pd.merge(home_games, away_games, on='GAME_ID')
model_df['HOME_WIN'] = (model_df['HOME_WL'] == 'W').astype(int)



#games['GAME_DATE'] = pd.to_datetime(games['GAME_DATE'])
#games = games.sort_values(['TEAM_ID','GAME_DATE'])



model_df.columns
```

```python
star_map = dict(zip(stars_list['TEAM_ID'], stars_list['PLAYER_ID']))
model_df['HOME_STAR_ID'] = model_df['HOME_TEAM_ID'].map(star_map)
model_df['AWAY_STAR_ID'] = model_df['AWAY_TEAM_ID'].map(star_map)

# 1. Create the set for instant lookup (same as before)
active_pairs = set(zip(star_attendance['GAME_ID'], star_attendance['PLAYER_ID']))
```

```python
for col in ['AST', 'REB', 'TOV','TS%','NET_RAT']:
    model_df[f'DIFF_{col}'] = model_df[f'HOME_ROLL_{col}'] - model_df[f'AWAY_ROLL_{col}']
model_df['DIFF_SEASON_WIN_PCT'] = model_df['HOME_SEASON_WIN_PCT'] - model_df['AWAY_SEASON_WIN_PCT']
model_df['DIFF_REST'] = model_df['HOME_DAYS_REST'] - model_df['AWAY_DAYS_REST']
model_df['DIFF_STAR_ACTIVE'] = [
    (1 if (game, h_star) in active_pairs else 0) - (1 if (game, a_star) in active_pairs else 0)
    for game, h_star, a_star in zip(model_df['GAME_ID'], model_df['HOME_STAR_ID'], model_df['AWAY_STAR_ID'])
]
model_df['DIFF_B2B'] = model_df['HOME_IS_B2B'] - model_df['AWAY_IS_B2B']
model_df['DIFF_GAMES_4D'] = model_df['HOME_GAMES_IN_LAST_4_DAYS'] - model_df['AWAY_GAMES_IN_LAST_4_DAYS']
model_df['DIFF_GAMES_5D'] = model_df['HOME_GAMES_IN_LAST_5_DAYS'] - model_df['AWAY_GAMES_IN_LAST_5_DAYS']
model_df['DIFF_3_IN_4'] = model_df['HOME_FATIGUE_3_IN_4'] - model_df['AWAY_FATIGUE_3_IN_4']
model_df['DIFF_4_IN_5'] = model_df['HOME_FATIGUE_4_IN_5'] - model_df['AWAY_FATIGUE_4_IN_5']
model_df['DIFF_ADJ_NET_RAT'] = model_df['HOME_ADJ_NET_RAT'] - model_df['AWAY_ADJ_NET_RAT']

model_df['HOME_FATIGUE_PENALTY'] = (
    1
    - 0.12 * model_df['HOME_IS_B2B']
    - 0.08 * model_df['HOME_FATIGUE_3_IN_4']
    - 0.05 * model_df['HOME_FATIGUE_4_IN_5']
    - 0.05 * model_df['HOME_GAMES_IN_LAST_5_DAYS'].clip(lower=2) / 3
)

model_df['AWAY_FATIGUE_PENALTY'] = (
    1
    - 0.12 * model_df['AWAY_IS_B2B']
    - 0.08 * model_df['AWAY_FATIGUE_3_IN_4']
    - 0.05 * model_df['AWAY_FATIGUE_4_IN_5']
    - 0.05 * model_df['AWAY_GAMES_IN_LAST_5_DAYS'].clip(lower=2) / 3
)

model_df['DIFF_FATIGUE_ADJ_NET_RAT'] = (
    model_df['HOME_ADJ_NET_RAT'] * model_df['HOME_FATIGUE_PENALTY']
    - model_df['AWAY_ADJ_NET_RAT'] * model_df['AWAY_FATIGUE_PENALTY']
)


diff_features = [col for col in model_df.columns if 'DIFF_' in col]

model_df['GAME_DATE'] = pd.to_datetime(model_df['GAME_DATE'])
model_df = model_df.sort_values('GAME_DATE').reset_index(drop=True)

X = model_df[diff_features]
y = model_df['HOME_WIN']


split_index = int(len(model_df)*0.8)
X_train, X_test = X.iloc[:split_index], X.iloc[split_index:]
y_train, y_test = y.iloc[:split_index], y.iloc[split_index:]

print(f'Training on {len(X_train)} games, testing on {len(X_test)} games')
print(model_df[:20])
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV

param_grid = {'max_depth':[3,5,7], 
              'learning_rate':[0.01,0.1,0.2],
              'n_estimators': [100, 200],
              'subsample': [0.8, 1.0]
              }

xgb_model = XGBClassifier(eval_metric='logloss')
grid_search = GridSearchCV(xgb_model, param_grid, cv=5, n_jobs=-1)
grid_search.fit(X_train, y_train)


print(f"Best Parameters: {grid_search.best_params_}")
print(f"Best Score from Grid Search: {grid_search.best_score_}")
```

```python
best_model = grid_search.best_estimator_
train_probs = best_model.predict_proba(X_train)[:,1]
train_y = y_train
```

```python
from sklearn.isotonic import IsotonicRegression
iso = IsotonicRegression(out_of_bounds='clip')
iso.fit(train_probs, train_y)
xgb_model.fit(X_train,y_train)
```

```python
test_probs_raw = xgb_model.predict_proba(X_test)[:,1]
test_probs_cal = iso.transform(test_probs_raw)
print(test_probs_cal)
```
